```python
Gasabo = df[df["district"] == 'Gasabo'].shape[0]
print("Gasabo : ", Gasabo)
Kicukiro = df[df["district"] == 'Kicukiro'].shape[0]
Nyarugenge = df[df["district"] == 'Nyarugenge'].shape[0]
Rusizi = df[df["district"] == 'Rusizi'].shape[0]
print("Kicukiro : ",Kicukiro)
print("Rusizi : ",Rusizi)
```

```python
#For top student
score = df['score'].max()
position = df['score'].idxmax()
name = df['name'][position]
print("The top student is ",name," with a score of ",score)
```

```python
#for lowest
score = df['score'].min()
position = df['score'].idxmin()
name = df['name'][position]
print("The lowest student is ",name," with a score of ",score)
```
