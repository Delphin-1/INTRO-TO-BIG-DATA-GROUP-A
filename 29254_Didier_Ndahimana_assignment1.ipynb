```python
#PART 5:

#EXERCISE 5.1

# 1

import pandas as pd

df=pd.read_excel("week1_students.xlsx")


print( "the total number of students is: ", len(df))



```

```python
#2

f=round(df["score"].mean(),1)
print( "the average score is: ", f)

```

```python
#3

maximum = df["score"].max()
print("the maximum score is: ", maximum)

minimum = df["score"].min()
print("the minimum score is: ", minimum)



```

```python
#4

range = maximum - minimum
print("the range of scores is: ", range)


```

```python
#EXERCISE 5.2

#1

print(df.iloc[0])

print(df.iloc[19])

```

```python
#2

names=df.iloc[5]["name"]
districts=df.iloc[5]["district"]
programs=df.iloc[5]["program"]
scores=df.iloc[5]["score"]

print(f"{names} from {districts} studies {programs} and scored {scores}")





```

```python
#EXERCISE 5.3

#1

gsb=df[df["district"]=="Gasabo"]
kckr=df[df["district"]=="Kicukiro"]
rwg=df[df["district"]=="Rwamagana"]

print("students from Gasabo is: ", len(gsb))
print("students from Kicukiro is: ", len(kckr))
print("students from Rwamagana is: ", len(rwg))


```

```python
#EXERCISE 5.4

#1

first_winner=df[df["score"]==df["score"].max()]

print(f"student is {first_winner.iloc[0]['name']},score:{first_winner.iloc[0]['score']}")
```

```python
#BONUS

lowest_winner=df[df["score"]==df["score"].min()]
print(f"student is{lowest_winner.iloc[0]['name']},score:{lowest_winner.iloc[0]['score']}")  
```

```python
first_winner_attendance=df[df["score"]==df["score"].max()]["attendance_pct"].values[0]
print(f"the top student attendance is  {first_winner_attendance}")

lowest_winner_attendance=df[df["score"]==df["score"].min()]["attendance_pct"].values[0]
print(f"the lowest student attendance is  {lowest_winner_attendance}")
```
