```python
course = "Big Data Analytics"

# (a) Convert to UPPERCASE
print(course.upper())
# Output: BIG DATA ANALYTICS

# (b) Count characters
print(len(course))
# Output: 18

# (c) Get first 3 characters (slicing)
print(course[0:3])
# Output: Big
```

```python
#LIST OPERATION

scores = [72, 85, 91, 64, 78]

# (a) Print first and last score
print("First score:", scores[0])
print("Last score:", scores[-1])

# (b) Print number of scores
print("Number of scores:", len(scores))

# (c) Print the average
average = sum(scores) / len(scores)
print("Average:", average)

# (d) Print highest and lowest
print("Highest score:", max(scores))
print("Lowest score:", min(scores))
```

```python
#Adding a new key, modifying an existing value, and viewing all keys and the whole dictionary

me = {
    "name": "JASON",
    "age": 20,
    "program": "Information Systems",
    "district": "Kicukiro"
}

# (a) Add a new key "score" with a value
me["score"] = 85

# (b) Increase age by 1
me["age"] = me["age"] + 1

# (c) Print all the keys
print("Keys:", me.keys())

# Print the whole dictionary
print("Full dictionary:", me)
```
