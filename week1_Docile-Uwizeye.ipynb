```python
scores = [72, 85, 91, 64, 78]

# (a) First and last score
print(scores[0])
print(scores[-1])

# (b) Number of scores
print(len(scores))

# (c) Average
print(sum(scores) / len(scores))

# (d) Highest and lowest score
print(max(scores))
print(min(scores))
```

```python
scores = [72, 85, 91, 64, 78]

# (a) Append 88
scores.append(88)

# (b) Replace position 3 (64) with 66
scores[3] = 66

# (c) Print updated list and new average
print(scores)
print(sum(scores) / len(scores))
```
