```python
import numpy as np
import pandas as pd

df = pd.read_excel("week5_kigali_weather.xlsx")

dates = pd.to_datetime(df["date"], unit="D", origin="1899-12-30").dt.strftime("%Y-%m-%d").to_numpy()
temps = df["temp_c"].to_numpy()
rain  = df["rainfall_mm"].to_numpy()
hum   = df["humidity_pct"].to_numpy()

print(type(temps), len(temps), dates[0], dates[-1]) 
```

```python
import numpy as np

print(f"Mean: {temps.mean():.2f} °C")  
print(f"Max:  {temps.max():.1f} °C")
print(f"Min:  {temps.min():.1f} °C")
print(f"Std:  {temps.std():.2f} °C")

pos = np.argmax(temps)
print(f"Hottest day: {dates[pos]} at {temps[pos]} °C")

hot_mask = temps >= 26
print(f"Hot days: {hot_mask.sum()}")
print(f"Avg temp on hot days: {temps[hot_mask].mean():.2f} °C")

rainy_mask = rain > 0
print(f"Rainy days: {rainy_mask.sum()}")
print(f"Total rainfall: {rain.sum():.1f} mm")
print(f"Heaviest day: {rain.max():.1f} mm")

print(f"Avg temp, rainy days: {temps[rain > 0].mean():.2f} °C")
print(f"Avg temp, dry days:   {temps[rain == 0].mean():.2f} °C")
print(f"Hot AND rainy days: {((temps >= 26) & (rain > 0)).sum()}")
print(f"Days with humidity >= 70: {(hum >= 70).sum()}")

anomaly = temps - temps.mean()
print("Days > 3 °C above mean:", dates[anomaly > 3])
print(f"Days > 3 °C below mean: {(anomaly < -3).sum()}")

first, last = temps[:30].mean(), temps[30:].mean()
print(f"First 30 days: {first:.2f} °C | Last 30 days: {last:.2f} °C")
```

```python
wettest = sorted(zip(dates, rain), key=lambda p: p[1], reverse=True)[:5]
for d, r in wettest:         
    print(f"{d}: {r} mm")
```
