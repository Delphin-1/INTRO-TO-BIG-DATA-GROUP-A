```python
# Exercise 1.1 — The Grading Ladder

scores = [91, 76, 64, 52, 43]

for score in scores:
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"

    print(f"Score {score} -> Grade {grade}")
```

```python
# Exercise 1.2 — Combined Conditions

def validate(score, attendance):
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")


validate(76, 92)
validate(76, 68)
validate(43, 95)
```

```python
# Exercise 2.2 — Accumulator Pattern

scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]
total = 0
count = 0
passed = 0
failed = 0
for score in scores:
    total += score
    count += 1

    if score >= 50:
        passed += 1
    else:
        failed += 1

average = total / count

print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

```python
# Exercise 2.3 — While Loop

savings = 0
monthly_saving = 15000
target = 250000
months = 0

while savings < target:
    savings += monthly_saving
    months += 1

print("Months needed:", months)
print("Final savings:", savings, "RWF")
```

```python
# Exercise 3.1 — get_grade function

def get_grade(score):
    """Return the letter grade for a score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

```python
# Exercise 3.2 — pass_rate function

def pass_rate(score_list):
    passed = 0

    for score in score_list:
        if score >= 50:
            passed += 1

    return (passed / len(score_list)) * 100


scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print(pass_rate(scores))
```

```python
with open("week3_students.csv", "r", encoding="utf-8-sig") as f:
    lines = f.readlines()

print("Number of lines:", len(lines))
print("Header:", lines[0])
print("First student:", lines[1])
```

```python
# Exercise 4.2 — Extract student names

names = []

for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Number of students:", len(names))
print("First 5 names:", names[:5])
```

```python
# Exercise 5.1 — Resilient data pipeline

total = 0
count = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")

    try:
        score = int(parts[7])
        total += score
        count += 1

    except ValueError:
        bad += 1
        bad_ids.append(parts[0])

average = total / count

print("Valid scores:", count)
print("Bad scores:", bad)
print("Bad IDs:", bad_ids)
print("Average:", round(average, 1))
```

```python
# Exercise 5.2 — Passed/Failed and Best Student

passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")

    try:
        score = int(parts[7])

        if score >= 50:
            passed += 1
        else:
            failed += 1

        if score > best_score:
            best_score = score
            best_name = parts[1]

    except ValueError:
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Top score:", best_score)
print("Top student:", best_name)
print("Grade:", get_grade(best_score))
```

```python
# Exercise 5.3 — District averages

sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    district = parts[5]

    try:
        score = int(parts[7])

        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1

    except ValueError:
        continue

for district in sums:
    average = sums[district] / counts[district]
    print(district, ":", round(average, 1))
```

```python
# Bonus — Save report to a text file

with open("week3_report.txt", "w") as f:
    f.write("WEEK 3 STUDENT REPORT\n")
    f.write("=====================\n")
    f.write(f"Valid scores: {count}\n")
    f.write(f"Bad scores: {bad}\n")
    f.write(f"Average score: {round(average, 1)}\n")
    f.write(f"Bad IDs: {bad_ids}\n")
    f.write(f"Passed: {passed}\n")
    f.write(f"Failed: {failed}\n")
    f.write(f"Top student: {best_name}\n")
    f.write(f"Top score: {best_score}\n")
    f.write(f"Top grade: {get_grade(best_score)}\n")

print("Report saved successfully.")
```
