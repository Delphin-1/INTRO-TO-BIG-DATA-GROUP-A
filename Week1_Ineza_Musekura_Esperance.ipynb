```python
import pandas as pd
students = pd.read_excel("week1_students.xlsx")

print(students)
```

```python
print(type(students))
```

```python
students = students.to_dict(orient="records")
print(type(students))
```

```python
print(students[0])
```

```python
all_scores = [s["score"] for s in students]

print(f"Number of students: {len(students)}")
print(f"Class average score: {round(sum(all_scores) / len(all_scores), 1)}")
print(f"Highest score: {max(all_scores)}")
print(f"Lowest score: {min(all_scores)}")
print(f"Score range: {max(all_scores) - min(all_scores)}")
```

```python
print(students[0])
print(students[-1])

student = students[5]
print(f"{student['name']} from {student['district']} studies {student['program']} and scored {student['score']}.")
```

```python
districts = [s["district"] for s in students]

print(f"Gasabo: {districts.count('Gasabo')}")
print(f"Kicukiro: {districts.count('Kicukiro')}")
print(f"Nyarugenge: {districts.count('Nyarugenge')}")
```

```python
top_score = max(all_scores)
top_index = all_scores.index(top_score)

print(f"Top student: {students[top_index]['name']}")
print(f"Top score: {top_score}")
```

```python
lowest_score = min(all_scores)
lowest_index = all_scores.index(lowest_score)

print(f"Lowest student: {students[lowest_index]['name']}")
print(f"Lowest score: {lowest_score}")
print(f"Attendance: {students[lowest_index]['attendance_pct']}%")
```

6.1 Is this 20-student dataset big data?
=> No, this 20-student dataset is not Big Data. It is a small dataset that can easily be stored, processed, and analyzed using basic Python tools. The 5 V's of Big Data are Volume, Velocity, Variety, Veracity, and Value. In this dataset, the 5 V's are not clearly present at a Big Data scale because there are only 20 records, the data is mostly structured, and there is no continuous high-speed data generation or large-scale complexity.
