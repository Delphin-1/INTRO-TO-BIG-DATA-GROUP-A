## Part 6 · Reflection

**1. Is this 20-student dataset Big Data? Which of the 5 V's does it clearly not have yet?**

No — it is not Big Data. It has small volume (only 20 rows, a few KB), modest variety (structured rows of text and numbers only), and high veracity (clean, complete records). The V it clearly lacks is Volume (and, to a lesser degree, Velocity, since nothing is streaming). Big Data typically involves terabytes-to-petabytes arriving continuously, which a 20-row CSV cannot represent.
