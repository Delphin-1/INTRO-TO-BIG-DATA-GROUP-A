```python
# ===== Exercise 3.1 =====
scores = [72, 85, 91, 64, 78]

print(scores[0])                 # (a) first score
print(scores[-1])                # (a) last score
print(len(scores))               # (b) number of scores
print(sum(scores) / len(scores)) # (c) average
print(max(scores))               # (d) highest
print(min(scores))               # (d) lowest
```

```python
# ===== Exercise 3.2 =====
scores.append(88)     # (a) adds 88 to the end
scores[3] = 66          # (b) replaces position 3 (4th item) with 66
print(scores)
print(sum(scores) / len(scores))  # new average
```
