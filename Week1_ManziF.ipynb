```python
# PART VI - Reflection

# 1. Is this 20-student dataset Big Data?
# No, this is NOT Big Data. It clearly lacks VOLUME — 20 records is tiny.
# It also lacks VELOCITY (it is static, not a live stream) and VARIETY
# (all data is neatly structured in one table with no unstructured sources).
# Of the 5 V's, Volume is the most obviously missing one.

# 2. One thing that surprised or confused me:
# I was surprised that Python's + operator on strings does not add numbers —
# it just glues the text together. I expected "3" + "4" to give 7,
# but it gives "34" instead. This made me realise how important data types
# are: always convert text to int/float before doing any arithmetic.
```
