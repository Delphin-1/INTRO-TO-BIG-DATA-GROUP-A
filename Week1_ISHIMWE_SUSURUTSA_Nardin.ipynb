### Question 1.3 — Think about it (1 mark)
What does `"3" + "4"` produce? Run it, then explain in one sentence why the answer is not 7.

```python
result = "3" + "4"
print(result)
```

**Explanation:** `"3"` and `"4"` are **strings**, not integers, so the `+` operator performs **string concatenation** (joining the text side by side) instead of numeric addition — Python only adds numbers with `+`; with two strings it glues them together, producing `"34"` rather than `7`.

### Exercise 3.1 (4 marks)

```python
scores = [72, 85, 91, 64, 78]

print(scores[0], scores[-1])            # (a) first and last score
print(len(scores))                      # (b) number of scores
print(sum(scores) / len(scores))        # (c) average
print(max(scores), min(scores))         # (d) highest and lowest score
```

## Part 6 · Reflection

**1. Is this 20-student dataset Big Data? Which of the 5 V's does it clearly not have yet?**

No — 20 records is a tiny, everyday-sized dataset that fits easily in memory and can be scanned instantly, so it is not "Big Data" in the technical sense. Of the 5 V's (Volume, Velocity, Variety, Veracity, Value), it most clearly lacks **Volume**: real big-data workloads involve millions or billions of records, not 20. It also has essentially no **Velocity** — the data is static and was not generated or updated continuously in real time.

**2. Describe one thing from today's session that surprised or confused you.**

One thing that stood out was how `"3" + "4"` produces `"34"` instead of `7` — it's a small detail, but it's a good reminder that data loaded from files (like CSVs) often arrives as text even when it "looks like" a number, so it's easy to get silently wrong results (or errors) if you forget to convert types before doing arithmetic.
