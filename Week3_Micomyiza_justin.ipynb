```python
# part 4 reading the file
# exercises 4.1 first contact
from openpyxl import load_workbook

# Open the Excel file
workbook = load_workbook("week3_students.xlsx", data_only=True)
sheet = workbook.active

# Read all rows
rows = list(sheet.iter_rows(values_only=True))

print("Number of lines:", len(rows))
print("Header line:", ", ".join(str(x) for x in rows[0]))
print("First student's line:", ", ".join(str(x) for x in rows[1]))

# There are 41 lines because there are 40 student records plus 1 header line.
```

```python
#exercise 4.2 extract a column 
names = []

for row in rows[1:]:
  names.append(row[1])

print("Count:", len(names))
print("First 5 names:", names[:5])
```

```python
#part 5 
#exercises 5.1
# Exercise 5.1 — Survive the mess

total = 0
valid = 0
bad = 0
bad_ids = []

for row in rows[1:]:
    student_id = row[0]
    score_value = row[7]

    try:
        score = int(score_value)
        total += score
        valid += 1

    except (ValueError, TypeError):
        bad += 1
        bad_ids.append(student_id)

average = total / valid if valid > 0 else 0

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", f"{average:.1f}")
```

```python
#exercises 5.2 
# Exercise 5.2 — Grades & the top student
def get_grade(score):
    """Return the letter grade for a numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"
passed = 0
failed = 0

best_score = -1
best_name = ""

for row in rows[1:]:
    name = row[1]
    score_value = row[7]

    try:
        score = int(score_value)

    except (ValueError, TypeError):
        continue

    # Count passed and failed
    if score >= 50:
        passed += 1
    else:
        failed += 1

    # Find highest score
    if score > best_score:
        best_score = score
        best_name = name

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name)
print("Top score:", best_score)
print("Top student's grade:", get_grade(best_score))
```

```python


sums = {}
counts = {}

for row in rows[1:]:
    district = row[5]
    score_value = row[7]

    try:
        score = int(score_value)

    except (ValueError, TypeError):
        continue

    # Accumulate total score for each district
    sums[district] = sums.get(district, 0) + score

    # Accumulate number of valid students
    counts[district] = counts.get(district, 0) + 1


# Calculate and print every district's average
highest_district = ""
highest_average = -1

for district in sums:
    average = sums[district] / counts[district]

    print(f"{district}: {average:.1f}")

    if average > highest_average:
        highest_average = average
        highest_district = district


print()
print("District with highest average:", highest_district)
print("Highest district average:", f"{highest_average:.1f}")
```
