```python
def pass_rate(score_list):
    passed = 0
    total = 0
    for s in score_list:
        total = total + 1
        if s >= 50:
            passed = passed + 1
    return (passed / total) * 100

print(pass_rate(scores))
```

```python
import pandas as pd
df = pd.read_excel("week3_students.xlsx")
df.to_csv("week3_students.csv",index=False)
```

```python
with open("week3_students.csv", "r") as f:
    lines = f.readlines()

print(len(lines))
print(lines[0])
print(lines[1])
```

```python
names = []
for line in lines[1:]:               
    parts = line.strip().split(",")
    names.append(parts[1])            

print("Count:", len(names))           
print("First 5 names:", names[:5])
```

```python
total = 0
count_valid = 0
count_bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        count_valid = count_valid + 1
    except ValueError:
        count_bad = count_bad + 1
        bad_ids.append(parts[0])

average_valid = total / count_valid
```

```python
total = 0
count_valid = 0
count_bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
        total = total + score
        count_valid = count_valid + 1
    except ValueError:
        count_bad = count_bad + 1
        bad_ids.append(parts[0])

average_valid = total / count_valid

print("Valid:", count_valid)
print("Bad:", count_bad)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", round(average_valid, 1))
```

```python
def get_grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"
```

```python
passed = 0
failed = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue                    

    if score >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

    if score > best_score:          
        best_score = score
        best_name = parts[1]

print("Passed:", passed, "Failed:", failed)
print("Top student:", best_name, "with score", best_score, "-> grade", get_grade(best_score))
```

```python
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        s = int(parts[7])
    except ValueError:
        continue
    d = parts[5]
    sums[d] = sums.get(d, 0) + s
    counts[d] = counts.get(d, 0) + 1

for d in sums:
    print(d, ":", round(sums[d] / counts[d], 1))
```

```python
with open("week3_report.txt", "w") as f:
    f.write("Week 3 Student Report\n")
    f.write(f"Total students: {len(names)}\n")
    f.write(f"Valid records: {count_valid}\n")
    f.write(f"Bad records: {count_bad} {bad_ids}\n")
    f.write(f"Average score (valid only): {round(average_valid, 1)}\n")
    f.write(f"Top student: {best_name} ({best_score}, grade {get_grade(best_score)})\n")

with open("week3_report.txt") as f:
    print(f.read())
```
