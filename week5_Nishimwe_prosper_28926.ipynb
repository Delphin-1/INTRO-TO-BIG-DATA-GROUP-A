```python
import time

arr = np.arange(1_000_000)

start = time.time()
loop_result = [x * 2 for x in arr]
loop_time = time.time() - start

start = time.time()
numpy_result = arr * 2
numpy_time = time.time() - start

print(f"Python loop time: {loop_time:.6f} seconds")
print(f"NumPy time: {numpy_time:.6f} seconds")
print(f"NumPy was approximately {loop_time / numpy_time:.1f} times faster")
```

Part 4 — Boolean Masking

```python
hot_mask = temps >= 26

print(f"Number of hot days: {hot_mask.sum()}")
print(f"Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
```

```python
rain_mask = rain > 0

print(f"Number of rainy days: {rain_mask.sum()}")
print(f"Total rainfall: {rain.sum():.1f} mm")
print(f"Heaviest single-day rainfall: {rain.max():.1f} mm")
```

Exercise 4.3(a) — Rainy vs dry

```python
rain_mask = rain > 0

print(f"Average temperature on rainy days: {temps[rain_mask].mean():.2f} °C")
print(f"Average temperature on dry days: {temps[~rain_mask].mean():.2f} °C")
```

Exercise 4.3(b) — Hot AND rainy

```python
hot_and_rainy = (temps >= 26) & (rain > 0)

print(f"Hot and rainy days: {hot_and_rainy.sum()}")
```

Exercise 4.3(c) — Humidity ≥ 70%

```python
humidity_mask = hum >= 70

print(f"Days with humidity >= 70%: {humidity_mask.sum()}")
```

Part 5 — Anomalies & Trends

```python
anomaly = temps - temps.mean()

above_mask = anomaly > 3

print("Dates more than 3 °C above the mean:")
print(dates[above_mask])

below_mask = anomaly < -3

print(f"Days more than 3 °C below the mean: {below_mask.sum()}")
```

```python
first_30 = temps[:30]
last_30 = temps[30:]

print(f"Average temperature, first 30 days: {first_30.mean():.2f} °C")
print(f"Average temperature, last 30 days: {last_30.mean():.2f} °C")

# Kigali got cooler because the average temperature decreased.
```

Top 5 wettest days

```python
wettest_days = sorted(
    zip(dates, rain),
    key=lambda x: x[1],
    reverse=True
)

for date, rainfall in wettest_days[:5]:
    print(f"{date}: {rainfall} mm")
```

### Reflection

**1. What did NumPy hide from you , and why was it still important to learn the loop version first?**

NumPy hid the repetitive process of going through each value one by one. Instead of writing a loop to calculate an average, I could simply use `temps.mean()`. Learning loops first was important because it helped me understand the basic logic behind the calculation. NumPy then made the same operations shorter, faster, and easier to apply to large datasets.

**2. One insight about Kigali's weather**

The analysis revealed a clear cooling trend during the 60 days. The average temperature during the first 30 days was about 26.21 °C, while the average during the last 30 days was about 21.78 °C. The hottest day was 17/09/2026, with a temperature of 28.6 °C.
