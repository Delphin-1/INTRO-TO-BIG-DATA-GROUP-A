## Part 6 · Reflection

**1. Is this 20-student dataset Big Data? Which of the 5 V's does it clearly not have yet?**

No, a 20 student dataset is not Big Data. It clearly lacks **Volume**  Big Data typically refers to datasets so large (gigabytes, terabytes, or more) that they cannot be processed with standard tools on a single computer. A 20-row dataset fits comfortably in memory and can be analyzed by hand or with a simple spreadsheet, so it doesn't meet the scale threshold that "Volume" requires. It may hint at other V's (e.g., Variety, since it mixes text, numbers, and categories), but Volume is the most obviously missing dimension.

**2. Describe one thing from today's session that surprised or confused you.**

One thing that stood out was how `"3" + "4"` returns `"34"` instead of `7`  a small reminder that Python treats data literally by type, and that real-world datasets (like `age_text = "21"`) often store numbers as text, which means a "number" isn't always ready for math until it's explicitly converted. It's an easy mistake to make when moving quickly through a dataset without checking each column's type first.
