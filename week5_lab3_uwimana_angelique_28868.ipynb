### exercise 5.2

```python
midpoint = len(temps) // 2

first_half_avg = temps[:midpoint].mean()
last_half_avg = temps[midpoint:].mean()

print(f"Average temperature for first half: {first_half_avg:.2f} °C")
print(f"Average temperature for last half: {last_half_avg:.2f} °C")

if last_half_avg > first_half_avg:
    print("# Kigali got warmer.")
elif last_half_avg < first_half_avg:
    print("# Kigali got cooler.")
else:
    print("# The average temperature stayed the same.")
```
