## 0. Load and Prepare the Weather Dataset
First, we import the required libraries and load the provided Kigali weather data. The date column is converted to a readable format when necessary.

```python
import numpy as np
import pandas as pd

# Load the provided Kigali weather dataset
weather = pd.read_excel("week5_kigali_weather.xlsx")

# Convert Excel date serial numbers into calendar dates when needed
if pd.api.types.is_numeric_dtype(weather["date"]):
    weather["date"] = pd.to_datetime(
        weather["date"], unit="D", origin="1899-12-30"
    )

# Convert the required columns into NumPy arrays for analysis
dates = weather["date"].astype(str).to_numpy()
temps = weather["temp_c"].to_numpy(dtype=float)
rain = weather["rainfall_mm"].to_numpy(dtype=float)
hum = weather["humidity_pct"].to_numpy(dtype=float)

print(f"Dataset loaded successfully: {len(weather)} records")
print(f"Columns: {', '.join(weather.columns)}")
```

## Part 1 — List Comprehensions (6 marks)
List comprehensions provide a concise way to transform and filter collections.

```python
# Sample temperatures for Part 1
temps_part1 = [24.4, 25.9, 28.6, 19.3, 26.0, 22.1]

# Exercise 1.1 — Convert Celsius to Fahrenheit
fahrenheit = [t * 9 / 5 + 32 for t in temps_part1]
print("Temperatures in Fahrenheit:", fahrenheit)

# Exercise 1.2(a) — Select temperatures of at least 26°C
hot = [t for t in temps_part1 if t >= 26]
print("Hot temperatures (>= 26°C):", hot)

# Exercise 1.2(b) — Select temperatures below 24°C and round them
cool_rounded = [round(t) for t in temps_part1 if t < 24]
print("Cool temperatures (< 24°C, rounded):", cool_rounded)

# Exercise 1.3 — Convert strings to floats and calculate the average
raw = ["24.4", "25.9", "28.6", "19.3"]
raw_floats = [float(value) for value in raw]
average_temperature = sum(raw_floats) / len(raw_floats)
    
print("Converted values:", raw_floats)
print(f"Average temperature: {average_temperature:.2f} °C")
```

## Part 2 — Lambda Functions (4 marks)
Lambda functions are useful for short, focused operations such as sorting and finding maximum values.

```python
# Exercise 2.1 — Sort results from highest to lowest and display the top 3
results = [
    ("Aline", 85),
    ("Eric", 78),
    ("Diane", 91),
    ("Samuel", 92),
    ("Grace", 76),
]

ranked = sorted(results, key=lambda item: item[1], reverse=True)
print("Top 3 results:", ranked[:3])

# Exercise 2.2 — Find the day with the highest rainfall
rain_log = [(3, 0.0), (7, 12.5), (11, 3.2), (14, 26.2), (18, 8.9)]
wettest = max(rain_log, key=lambda item: item[1])
print("Wettest day (day, rainfall_mm):", wettest)
```

### Exercise 3.2 — Describe the Temperatures
Use NumPy aggregation functions to summarize the temperature array.

```python
print(f"Mean temperature: {temps.mean():.2f} °C")
print(f"Maximum temperature: {temps.max():.2f} °C")
print(f"Minimum temperature: {temps.min():.2f} °C")
print(f"Standard deviation: {temps.std():.2f} °C")
```

### Exercise 3.3 — Identify the Hottest Day

```python
# np.argmax returns the position of the largest temperature value
hottest_index = np.argmax(temps)

print(f"Hottest date: {dates[hottest_index]}")
print(f"Hottest temperature: {temps[hottest_index]:.1f} °C")
```

## Part 4 — Boolean Masking (10 marks)
Boolean masks allow us to select only the observations that satisfy a condition.

### Exercise 4.1 — Hot Days

```python
# Select all days with temperatures of at least 26°C
hot_mask = temps >= 26
hot_day_count = hot_mask.sum()

print(f"Number of hot days (>= 26°C): {hot_day_count}")
print(f"Average temperature on hot days: {temps[hot_mask].mean():.2f} °C")
```

### Exercise 4.2 — Rainfall Report

```python
# A rainy day is any day with rainfall greater than zero
rainy_mask = rain > 0

print(f"Number of rainy days: {rainy_mask.sum()}")
print(f"Total rainfall: {rain.sum():.1f} mm")
print(f"Heaviest single-day rainfall: {rain.max():.1f} mm")
```

### Exercise 4.3 — Cross-Array Analysis

```python
# Compare temperatures across rainy and dry days
dry_mask = rain == 0

print(f"Average temperature on rainy days: {temps[rainy_mask].mean():.2f} °C")
print(f"Average temperature on dry days: {temps[dry_mask].mean():.2f} °C")

hot_and_rainy = (temps >= 26) & (rain > 0)
print(f"Hot AND rainy days: {hot_and_rainy.sum()}")
print(f"Days with humidity >= 70%: {(hum >= 70).sum()}")
```

## Part 5 — Anomalies & Trends (7 marks)
This section uses the mean temperature as a baseline for identifying unusual observations and comparing two halves of the dataset.

### Exercise 5.1 — Anomaly Detection

```python
# Calculate each day's difference from the overall mean temperature
anomaly = temps - temps.mean()

above_mask = anomaly > 3
below_mask = anomaly < -3

print("Dates more than 3°C above the mean:")
print(dates[above_mask])
print(f"Days more than 3°C below the mean: {below_mask.sum()}")
```

### Exercise 5.2 — Is the Season Changing?

```python
# Compare the first 30 days with the last 30 days
first30_avg = temps[:30].mean()
last30_avg = temps[30:].mean()

print(f"Average temperature, first 30 days: {first30_avg:.2f} °C")
print(f"Average temperature, last 30 days: {last30_avg:.2f} °C")

if last30_avg < first30_avg:
    print("Observation: the last 30 days were cooler than the first 30 days.")
elif last30_avg > first30_avg:
    print("Observation: the last 30 days were warmer than the first 30 days.")
else:
    print("Observation: both 30-day periods had the same average temperature.")
```

## Bonus — Top 5 Wettest Days (+3 marks)
Sort the dates by rainfall and display the five days with the highest rainfall.

```python
wettest_days = sorted(
    zip(dates, rain),
    key=lambda item: item[1],
    reverse=True,
)

print("Top 5 wettest days:")
for date, rainfall in wettest_days[:5]:
    print(f"{date}: {rainfall:.1f} mm")
```

### 1. What did NumPy hide, and why learn loops first?

NumPy hides many of the repeated steps required to process values one by one. For example, `temps.mean()` calculates the average of an entire array in one concise expression, while vectorized operations apply calculations to many values at once. Learning loops first is still important because loops explain the underlying ideas of repetition, accumulation, and step-by-step processing. This makes it easier to understand what concise NumPy operations are doing behind the scenes.
