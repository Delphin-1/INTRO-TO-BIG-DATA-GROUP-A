## Part 1 — Making Decisions (5 marks)
### Exercise 1.1 — The grading ladder (3 marks)

```python
def print_grade(score):
    if score >= 85:
        print("A")
    elif score >= 70:
        print("B")
    elif score >= 60:
        print("C")
    elif score >= 50:
        print("D")
    else:
        print("F")

for s in [91, 76, 64, 52, 43]:
    print_grade(s)
```

### Exercise 1.2 — Combined conditions (2 marks)

```python
def check_validation(score, attendance):
    if score >= 50 and attendance >= 75:
        print("Validated")
    elif score >= 50 and attendance < 75:
        print("Blocked: low attendance")
    else:
        print("Failed")

check_validation(76, 92)
check_validation(76, 68)
check_validation(43, 95)
```

### Question 1.3 — Think about it

If `elif score >= 50:` came **before** `elif score >= 70:`, every score that is actually 70 or higher (even 85+) would already satisfy `>= 50` first and stop there, so it would always be graded `D` instead of the correct higher letter — the ladder must go from strictest condition to loosest.

## Part 2 — Loops & the Accumulator (7 marks)

```python
scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]  # given for Part 2
```

### Exercise 2.1 — Loop + filter (2 marks)

```python
for s in scores:
    if s >= 80:
        print(f"Top score: {s}")
```

### Exercise 2.2 — The accumulator pattern (3 marks)

```python
total = 0
count = 0
passed = 0
failed = 0

for s in scores:
    total = total + s
    count = count + 1
    if s >= 50:
        passed = passed + 1
    else:
        failed = failed + 1

average = total / count
print("Total:", total)
print("Count:", count)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
```

### Exercise 2.3 — while loop (2 marks)

```python
saved = 0
monthly = 15000
target = 250000
months = 0

while saved < target:
    saved = saved + monthly
    months = months + 1

print(f"You need {months} months to afford the laptop.")
```

## Part 3 — Functions (6 marks)
### Exercise 3.1 — get_grade (3 marks)

```python
def get_grade(score):
    """Return the letter grade (A/B/C/D/F) for a given numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

for s in [91, 76, 43]:
    print(s, "->", get_grade(s))
```

### Exercise 3.2 — pass_rate (3 marks)

```python
def pass_rate(score_list):
    """Return the percentage of scores in score_list that are >= 50."""
    total = 0
    passed = 0
    for s in score_list:
        total = total + 1
        if s >= 50:
            passed = passed + 1
    return (passed / total) * 100

print("Pass rate:", pass_rate(scores))  # expected 90.0
```

## Part 5 — The Resilient Pipeline (12 marks)
### Exercise 5.1 — Survive the mess (4 marks)

```python
total_score = 0
valid = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    try:
        score = int(parts[7])
        total_score = total_score + score
        valid = valid + 1
    except ValueError:
        bad = bad + 1
        bad_ids.append(student_id)

avg_valid = total_score / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", round(avg_valid, 1))
```

### Exercise 5.2 — Grades & the top student (4 marks)

```python
passed_count = 0
failed_count = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue  # skip broken rows

    if score >= 50:
        passed_count = passed_count + 1
    else:
        failed_count = failed_count + 1

    if score > best_score:
        best_score = score
        best_name = parts[1]

print("Passed:", passed_count)
print("Failed:", failed_count)
print("Top student:", best_name, "-", best_score, "(" + get_grade(best_score) + ")")
```

### Exercise 5.3 — Stretch: report by district (4 marks)

```python
sums = {}
counts_dist = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]
    try:
        s = int(parts[7])
    except ValueError:
        continue  # skip broken rows

    sums[d] = sums.get(d, 0) + s
    counts_dist[d] = counts_dist.get(d, 0) + 1

best_district = ""
best_avg = -1
for d in sums:
    avg = sums[d] / counts_dist[d]
    print(f"{d}: {round(avg, 1)}")
    if avg > best_avg:
        best_avg = avg
        best_district = d

print(f"\nHighest average district: {best_district} ({round(best_avg, 1)})")
```

## Bonus (+3) — Your first output file

```python
report_lines = [
    f"Week 3 Report\n",
    f"Student count: {valid + bad}\n",
    f"Valid scores: {valid}\n",
    f"Bad scores: {bad}\n",
    f"Average score: {round(avg_valid, 1)}\n",
    f"Top student: {best_name} ({best_score}, grade {get_grade(best_score)})\n",
]

with open("week3_report.txt", "w") as f:
    for line in report_lines:
        f.write(line)

with open("week3_report.txt") as f:
    print(f.read())
```
