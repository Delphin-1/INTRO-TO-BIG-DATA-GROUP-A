### exercise 3.1 get grades

```python
def get_grade(score):
    """Return the letter grade for a score."""
    
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"


print(get_grade(91))
print(get_grade(76))
print(get_grade(43))
```

### exercise 3.2 pass_rate

```python
def pass_rate(score_list):
    """Return the percentage of scores that are 50 or higher."""
    
    total_scores = 0
    passed_scores = 0

    for score in score_list:
        total_scores += 1

        if score >= 50:
            passed_scores += 1

    return (passed_scores / total_scores) * 100


scores = [72, 85, 91, 64, 78, 47, 88, 55, 93, 61]

print("Pass rate:", pass_rate(scores))
```

## part4  reading the file
### exercise 4.1 first contact

```python
with open("week3_students.csv", "r") as f:  #[span_3](start_span)[span_3](end_span)
    lines = f.readlines()  #[span_4](start_span)[span_4](end_span)

print("Number of lines:", len(lines))  #[span_5](start_span)[span_5](end_span)
print("Header:", lines[0].strip())  #[span_6](start_span)[span_6](end_span)
print(f"First student line:{lines[1].strip()}")  #[span_7](start_span)[span_7](end_span)
```

## exercise 4.2 extract columns

```python
names = []

for line in lines[1:]:
    parts = line.strip().split(",")
    names.append(parts[1])

print("Number of names:", len(names))
print("First 5 names:", names[:5])
```

## part5 the resilient piprline
### exercise 5.1 survive the names

```python
total = 0
count = 0
bad = 0
bad_ids = []

for line in lines[1:]:
    parts = line.strip().split(",")
    
    # Skip empty or whitespace-only lines
    if not line.strip():
        continue
        
    try:
        score = int(parts[7])
        total += score
        count += 1
    except (IndexError, ValueError):
        bad += 1
        # Safely capture ID if present, otherwise note missing ID
        student_id = parts[0] if parts and parts[0] else "Unknown"
        bad_ids.append(student_id)

# Calculate average safely to avoid division by zero
average = total / count if count > 0 else 0

print("Valid scores:", count)
print("Bad scores:", bad)
print("Bad IDs:", bad_ids)
print("Average:", round(average, 1))
```

### exercise 5.2 grades and the top student

```python
passed = 0  #[span_4](start_span)[span_4](end_span)
failed = 0  #[span_5](start_span)[span_5](end_span)
best_score = -1  #[span_6](start_span)[span_6](end_span)
best_name = ""  #[span_7](start_span)[span_7](end_span)

for line in lines[1:]:  #[span_8](start_span)[span_8](end_span)
    # Skip blank lines
    if not line.strip():
        continue
        
    parts = line.strip().split(",")  #[span_9](start_span)[span_9](end_span)
    
    try:  #[span_10](start_span)[span_10](end_span)
        score = int(parts[7])  #[span_11](start_span)[span_11](end_span)
        
        if score >= 50:  #[span_12](start_span)[span_12](end_span)
            passed += 1  #[span_13](start_span)[span_13](end_span)
        else:  #[span_14](start_span)[span_14](end_span)
            failed += 1  #[span_15](start_span)[span_15](end_span)
            
        if score > best_score:  #[span_16](start_span)[span_16](end_span)
            best_score = score  #[span_17](start_span)[span_17](end_span)
            best_name = parts[1]  #[span_18](start_span)[span_18](end_span)
            
    except (ValueError, IndexError):  #[span_19](start_span)[span_19](end_span)[span_20](start_span)[span_20](end_span)
        continue  #[span_21](start_span)[span_21](end_span)

print("Passed:", passed)  #[span_22](start_span)[span_22](end_span)
print("Failed:", failed)  #[span_23](start_span)[span_23](end_span)
print("Top score:", best_score)  #[span_24](start_span)[span_24](end_span)
print("Top student:", best_name)  #[span_25](start_span)[span_25](end_span)
print("Grade:", get_grade(best_score))  #[span_26](start_span)[span_26](end_span)
```

## exercise 5.3

```python
sums = {}  #[span_9](start_span)[span_9](end_span)
counts = {}  #[span_10](start_span)[span_10](end_span)

for line in lines[1:]:  #[span_11](start_span)[span_11](end_span)
    # Skip empty lines
    if not line.strip():
        continue
        
    parts = line.strip().split(",")  #[span_12](start_span)[span_12](end_span)
    
    try:
        district = parts[5]  #[span_13](start_span)[span_13](end_span)
        score = int(parts[7])  #[span_14](start_span)[span_14](end_span)
        
        sums[district] = sums.get(district, 0) + score  #[span_15](start_span)[span_15](end_span)
        counts[district] = counts.get(district, 0) + 1  #[span_16](start_span)[span_16](end_span)
        
    except (ValueError, IndexError):  #[span_17](start_span)[span_17](end_span)[span_18](start_span)[span_18](end_span)
        continue  #[span_19](start_span)[span_19](end_span)

for district in sums:  #[span_20](start_span)[span_20](end_span)
    average = sums[district] / counts[district]  #[span_21](start_span)[span_21](end_span)
    print(district, ":", round(average, 1))  #[span_22](start_span)[span_22](end_span)
```
