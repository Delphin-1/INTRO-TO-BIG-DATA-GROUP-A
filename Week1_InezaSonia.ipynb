### Question 1.3 — Think about it (1 mark)

```python
print("3" + "4")
```

**Explanation:** `"3"` and `"4"` are **strings**, not numbers. The `+` operator on two strings performs **concatenation** (joining the text together), not arithmetic addition, so `"3" + "4"` produces the string `"34"` instead of the number `7`. To get `7` we would first need to convert both values to integers: `int("3") + int("4")`.

## Part 3 · Lists — Your First Descriptive Statistics (7 marks)

### Exercise 3.1 (4 marks)

```python
scores = [72, 85, 91, 64, 78]

print(scores[0])                    # (a) first score
print(scores[-1])                   # (a) last score
print(len(scores))                  # (b) number of scores
print(sum(scores) / len(scores))    # (c) average
print(max(scores))                  # (d) highest
print(min(scores))                  # (d) lowest
```

### Exercise 3.2 — Lists change (3 marks)

```python
scores.append(88)      # (a) new member joins with a score of 88
scores[3] = 66          # (b) position 3 (the 4th score) corrected to 66

print(scores)                       # (c) updated list
print(sum(scores) / len(scores))    # (c) new average
```
